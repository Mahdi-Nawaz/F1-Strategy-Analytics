# 2. Pace Analysis & Tyre Degradation

Exploring the data before building the dashboard.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

conn = sqlite3.connect('../data/f1_data.db')
laps = pd.read_sql('SELECT * FROM laps WHERE Year=2024 AND Race="Bahrain"', conn)
laps['LapTime'] = pd.to_numeric(laps['LapTime'], errors='coerce')
laps = laps.dropna(subset=['LapTime'])

# Filter outliers (pit stops etc.)
median = laps['LapTime'].median()
laps = laps[laps['LapTime'] < (median * 1.2)]

print(f"Loaded {len(laps)} clean laps")
```

```python
# Pace comparison: VER vs LEC
drivers_to_compare = ['VER', 'LEC']
compare_laps = laps[laps['Driver'].isin(drivers_to_compare)]

plt.figure(figsize=(12, 6))
sns.lineplot(data=compare_laps, x='LapNumber', y='LapTime', hue='Driver', marker='o')
plt.title('Lap Time Comparison: VER vs LEC')
plt.show()
```

```python
# Tyre Degradation per Stint
laps['TyreLife'] = pd.to_numeric(laps['TyreLife'], errors='coerce')
deg_laps = compare_laps.dropna(subset=['TyreLife', 'Compound'])

plt.figure(figsize=(12, 6))
sns.scatterplot(data=deg_laps, x='TyreLife', y='LapTime', hue='Compound', style='Driver')
plt.title('Tyre Degradation')
plt.show()
```
